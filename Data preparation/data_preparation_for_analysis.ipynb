```python
import pandas as pd

# === Load the full joined dataset ===
df_full = pd.read_csv("C:/Users/User/OneDrive - University of Exeter/Desktop/Business Project/olist_full_dataset.csv")

# === Step 1: Delivered orders only ===
df_seg = df_full[df_full['order_status'] == 'delivered'].copy()

# === Step 2: Select relevant columns ===
cols = [
    "customer_unique_id","order_id","order_purchase_timestamp",
    "payment_value","payment_installments",
    "product_category_name_english","review_score",
    "customer_city","customer_state", "review_comment_message"
]
df_seg = df_seg[[c for c in cols if c in df_seg.columns]].copy()

# === Step 3: Dates & light dtypes ===
df_seg["order_purchase_timestamp"] = pd.to_datetime(df_seg["order_purchase_timestamp"], errors="coerce")
for c in ["product_category_name_english","customer_city","customer_state"]:
    if c in df_seg.columns:
        df_seg[c] = df_seg[c].astype("category")

# === Step 4a: Fast numeric aggregations ===
num = df_seg.groupby("customer_unique_id").agg(
    total_orders=("order_id","nunique"),
    total_spend=("payment_value","sum"),
    avg_order_value=("payment_value","mean"),
    avg_payment_installments=("payment_installments","mean"),
    avg_review_score=("review_score","mean"),
    last_purchase=("order_purchase_timestamp","max"),
    category_diversity=("product_category_name_english", pd.Series.nunique)
    if "product_category_name_english" in df_seg.columns else ("order_id","size")
).reset_index()

# === Step 4b: Most common category ===
if "product_category_name_english" in df_seg.columns:
    cat_counts = (df_seg.dropna(subset=["product_category_name_english"])
                      .groupby(["customer_unique_id","product_category_name_english"])
                      .size().reset_index(name="cnt"))
    cat_top = (cat_counts.sort_values(["customer_unique_id","cnt"], ascending=[True,False])
                        .drop_duplicates("customer_unique_id")
                        .rename(columns={"product_category_name_english":"most_common_category"})
                        [["customer_unique_id","most_common_category"]])
else:
    cat_top = None

# === Step 4c: Most common city & state ===
city_top = None
state_top = None
if "customer_city" in df_seg.columns:
    city_counts = df_seg.groupby(["customer_unique_id","customer_city"]).size().reset_index(name="cnt")
    city_top = (city_counts.sort_values(["customer_unique_id","cnt"], ascending=[True,False])
                         .drop_duplicates("customer_unique_id")
                         [["customer_unique_id","customer_city"]])
if "customer_state" in df_seg.columns:
    state_counts = df_seg.groupby(["customer_unique_id","customer_state"]).size().reset_index(name="cnt")
    state_top = (state_counts.sort_values(["customer_unique_id","cnt"], ascending=[True,False])
                          .drop_duplicates("customer_unique_id")
                          [["customer_unique_id","customer_state"]])

# === Step 4d: Aggregate review comments (concatenate) ===
comments_agg = df_seg.groupby("customer_unique_id")["review_comment_message"] \
    .apply(lambda x: " | ".join(x.dropna().astype(str))) \
    .reset_index()

# === Step 5: Recency ===
latest_date = df_seg["order_purchase_timestamp"].max()
num["recency_days"] = (latest_date - num["last_purchase"]).dt.days
num = num.drop(columns=["last_purchase"])

# === Step 6: Merge everything ===
customer_seg = num
if cat_top is not None:
    customer_seg = customer_seg.merge(cat_top, on="customer_unique_id", how="left")
if city_top is not None:
    customer_seg = customer_seg.merge(city_top, on="customer_unique_id", how="left")
if state_top is not None:
    customer_seg = customer_seg.merge(state_top, on="customer_unique_id", how="left")

# Merge review comments
customer_seg = customer_seg.merge(comments_agg, on="customer_unique_id", how="left")

# === Step 7: Save ===
out_path = "C:/Users/User/OneDrive - University of Exeter/Desktop/Business Project/olist_customer_segmentation_dataset3_with_comments.csv"
customer_seg.to_csv(out_path, index=False)
print(out_path, customer_seg.shape)
```
